```python
from pathlib import Path

TRAIN_PATH = Path('../data/raw/train_images/train_images/')
TEST_PATH = Path('../data/raw/train_images/test_images/')

train_files = list(TRAIN_PATH.glob('*.jpg'))
test_files = list(TEST_PATH.glob('*.jpg'))

num_train_files = len(train_files)
num_test_files = len(test_files)

test_percent = 100 * num_test_files // (num_test_files + num_train_files)

print(f'There are {num_train_files} training images and {num_test_files} '
      f'test images\n\nThis makes a test size of {test_percent}%.')
```
